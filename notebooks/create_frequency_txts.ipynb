```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_terms_path = "../data/training_labels/Train/train_terms.tsv"
train_taxonomy_path = "../data/training_labels/Train/train_taxonomy.tsv"
```

```python
train_terms_df = pd.read_csv(train_terms_path, sep="\t")
train_taxonomy_df = pd.read_csv(train_taxonomy_path, sep="\t")
```

```python
display(train_terms_df)

## Get term counts
term_counts = train_terms_df["term"].value_counts()
print(term_counts)

# Plot term distribution
plt.figure(figsize=(10, 6))
term_counts.plot(kind="bar")
plt.title("Term Distribution in Training Data")
plt.xlabel("Term")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# filter out terms greater than 10
filtered_term_counts_10 = term_counts[term_counts > 10]
filtered_term_counts_10.to_csv("frequency_terms_10.csv", header=False)

# filter out terms greater than 50
filtered_term_counts_50 = term_counts[term_counts > 50]
filtered_term_counts_50.to_csv("frequency_terms_50.csv", header=False)
```
